```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
```

```python
import numpy as np
import pandas as pd
X = np.array([1.1,1.2,1.3,2.1,2.1,2.2])
Y = np.array([39000,39200,40000,np.nan,44300,45000])
df = pd.DataFrame({
    "Experience":X,
    "Salary":Y
})
print(df)
```

```python
plt.plot(X, Y, marker='D')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Original Data - Experience vs Salary")
plt.grid()
plt.show()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df["Salary"]=df["Salary"].fillna(df["Salary"].mean())
df
```

```python
df=df.drop_duplicates()
df
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
df
```

```python
plt.plot(df["Experience"],df["Salary"],marker="D")
plt.xlabel("Year Of Experience")
plt.ylabel("Salary")
plt.title("Original Data - Experience vs Salary")
plt.grid()
plt.show()
```

```python
X = df[["Experience"]]
Y = df["Salary"]
X_train, X_test, Y_train, Y_test=train_test_split(
    X, Y, test_size=0.2,random_state=1
)
```

```python
X_test.shape
```

```python
X_train.shape
```

```python
Y_test.shape
```

```python
Y_train.shape
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,Y_train)
```

```python
Y_pred=model.predict(X_test)
print("prediction value",Y_pred)
```

```python
r2=r2_score(Y_test,Y_pred)
print("R2 Score:",r2)
```

```python
experience =np.array([[5]])
salary=model.predict(experience)
print("Predicted Salary for % Year Experience:",salary[0])
```

```python
plt.plot(X,Y, label="Actual Salary")
plt.plot(X,model.predict(X),label="Regression Line")
plt.plot(df["Experience"],df["Salary"],marker="D")
plt.xlabel("Year Of Experience")
plt.ylabel("Salary")
plt.title("Original Data - Experience vs Salary")
plt.grid()
plt.show()
```
